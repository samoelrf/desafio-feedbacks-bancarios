```python
# ============================================================
# DETECÇÃO DE FRAUDES EM CARTÃO DE CRÉDITO
# VERSÃO CORRIGIDA PARA GOOGLE COLAB
# ============================================================

# ============================================================
# 1. IMPORTAR BIBLIOTECAS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_curve,
    auc,
    precision_recall_curve,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score
)

import xgboost as xgb
import shap

from imblearn.under_sampling import RandomUnderSampler
from imblearn.over_sampling import RandomOverSampler

import warnings
warnings.filterwarnings("ignore")


# ============================================================
# 2. CARREGAR DATASET
# ============================================================

url = "https://storage.googleapis.com/download.tensorflow.org/data/creditcard.csv"

df = pd.read_csv(url)

print("Dataset carregado com sucesso!")
print("Dimensões:", df.shape)

display(df.head())


# ============================================================
# 3. EXPLORAÇÃO
# ============================================================

print("\nInformações do dataset:")
df.info()

print("\nEstatísticas:")
display(df.describe())

print("\nValores ausentes:")
print(df.isnull().sum().sum())


# ============================================================
# 4. DESBALANCEAMENTO
# ============================================================

print("\nQuantidade por classe:")
print(df["Class"].value_counts())

print("\nPercentual por classe:")
print(df["Class"].value_counts(normalize=True) * 100)


plt.figure(figsize=(6, 4))

sns.countplot(
    x="Class",
    data=df
)

plt.title(
    "Distribuição das Classes (0 = Normal, 1 = Fraude)"
)

plt.xlabel("Classe")
plt.ylabel("Quantidade")

plt.show()


# ============================================================
# 5. CRIAR FEATURE Amount_log
# ============================================================

df["Amount_log"] = np.log1p(
    df["Amount"]
)

display(
    df[["Amount", "Amount_log"]].head()
)


# ============================================================
# 6. SEPARAR FEATURES E TARGET
# ============================================================

X = df.drop(
    "Class",
    axis=1
)

y = df["Class"]

print("\nX:", X.shape)
print("y:", y.shape)


# ============================================================
# 7. TRAIN / TEST
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("\nTreino:", X_train.shape)
print("Teste:", X_test.shape)


# ============================================================
# 8. STANDARD SCALER
# ============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
)

X_test_scaled = scaler.transform(
    X_test
)

print("\nStandardScaler aplicado.")


# ============================================================
# 9. REGRESSÃO LOGÍSTICA
# ============================================================

# CORREÇÃO:
# Aumentamos max_iter para evitar o
# ConvergenceWarning.

log_reg = LogisticRegression(
    max_iter=5000,
    solver="lbfgs",
    class_weight="balanced",
    random_state=42
)

log_reg.fit(
    X_train_scaled,
    y_train
)

y_pred_log = log_reg.predict(
    X_test_scaled
)

print("\n" + "=" * 60)
print("LOGISTIC REGRESSION")
print("=" * 60)

print(
    classification_report(
        y_test,
        y_pred_log,
        digits=4
    )
)


# ============================================================
# 10. RANDOM FOREST
# ============================================================

rf = RandomForestClassifier(
    n_estimators=100,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf.fit(
    X_train,
    y_train
)

y_pred_rf = rf.predict(
    X_test
)

print("\n" + "=" * 60)
print("RANDOM FOREST")
print("=" * 60)

print(
    classification_report(
        y_test,
        y_pred_rf,
        digits=4
    )
)


# ============================================================
# 11. XGBOOST
# ============================================================

# Calcular o peso da classe fraude

negativos = (y_train == 0).sum()
positivos = (y_train == 1).sum()

scale_pos_weight = negativos / positivos

print(
    "\nScale_pos_weight:",
    scale_pos_weight
)


# CORREÇÃO:
# Removemos use_label_encoder porque
# versões atuais do XGBoost não utilizam mais esse parâmetro.

xgb_model = xgb.XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,

    scale_pos_weight=scale_pos_weight,

    objective="binary:logistic",

    eval_metric="aucpr",

    random_state=42,
    n_jobs=-1
)

xgb_model.fit(
    X_train,
    y_train
)

y_pred_xgb = xgb_model.predict(
    X_test
)

print("\n" + "=" * 60)
print("XGBOOST")
print("=" * 60)

print(
    classification_report(
        y_test,
        y_pred_xgb,
        digits=4
    )
)


# ============================================================
# 12. COMPARAÇÃO
# ============================================================

modelos = {
    "Logistic Regression": y_pred_log,
    "Random Forest": y_pred_rf,
    "XGBoost": y_pred_xgb
}

print("\n" + "=" * 60)
print("COMPARAÇÃO DOS MODELOS")
print("=" * 60)

for nome, previsoes in modelos.items():

    print("\n", nome)

    print(
        classification_report(
            y_test,
            previsoes,
            digits=4
        )
    )


# ============================================================
# 13. MATRIZ DE CONFUSÃO
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred_xgb
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title(
    "Matriz de Confusão - XGBoost"
)

plt.xlabel("Predito")
plt.ylabel("Real")

plt.show()


# ============================================================
# 14. PROBABILIDADES
# ============================================================

prob_log = log_reg.predict_proba(
    X_test_scaled
)[:, 1]

prob_rf = rf.predict_proba(
    X_test
)[:, 1]

prob_xgb = xgb_model.predict_proba(
    X_test
)[:, 1]


# ============================================================
# 15. CURVA ROC
# ============================================================

fpr_log, tpr_log, _ = roc_curve(
    y_test,
    prob_log
)

fpr_rf, tpr_rf, _ = roc_curve(
    y_test,
    prob_rf
)

fpr_xgb, tpr_xgb, _ = roc_curve(
    y_test,
    prob_xgb
)

auc_log = auc(
    fpr_log,
    tpr_log
)

auc_rf = auc(
    fpr_rf,
    tpr_rf
)

auc_xgb = auc(
    fpr_xgb,
    tpr_xgb
)


plt.figure(figsize=(8, 6))

plt.plot(
    fpr_log,
    tpr_log,
    label=f"Logística - AUC {auc_log:.4f}"
)

plt.plot(
    fpr_rf,
    tpr_rf,
    label=f"Random Forest - AUC {auc_rf:.4f}"
)

plt.plot(
    fpr_xgb,
    tpr_xgb,
    label=f"XGBoost - AUC {auc_xgb:.4f}"
)

plt.plot(
    [0, 1],
    [0, 1],
    "--"
)

plt.xlabel(
    "False Positive Rate"
)

plt.ylabel(
    "True Positive Rate"
)

plt.title(
    "Curva ROC"
)

plt.legend()

plt.show()


# ============================================================
# 16. PRECISION-RECALL
# ============================================================

precision_curve, recall_curve, thresholds = (
    precision_recall_curve(
        y_test,
        prob_xgb
    )
)

average_precision = (
    average_precision_score(
        y_test,
        prob_xgb
    )
)


plt.figure(figsize=(8, 6))

plt.plot(
    recall_curve,
    precision_curve,
    label=f"XGBoost - AP {average_precision:.4f}"
)

plt.xlabel(
    "Recall"
)

plt.ylabel(
    "Precision"
)

plt.title(
    "Curva Precision-Recall"
)

plt.legend()

plt.show()


# ============================================================
# 17. TESTAR THRESHOLDS
# ============================================================

thresholds_testados = [
    0.10,
    0.20,
    0.30,
    0.40,
    0.50
]

print("\n" + "=" * 60)
print("ANÁLISE DE THRESHOLDS")
print("=" * 60)

for threshold in thresholds_testados:

    y_pred_threshold = (
        prob_xgb >= threshold
    ).astype(int)

    precision_value = precision_score(
        y_test,
        y_pred_threshold,
        zero_division=0
    )

    recall_value = recall_score(
        y_test,
        y_pred_threshold,
        zero_division=0
    )

    f1_value = f1_score(
        y_test,
        y_pred_threshold,
        zero_division=0
    )

    print(
        f"\nThreshold: {threshold:.2f}"
    )

    print(
        f"Precision: {precision_value:.4f}"
    )

    print(
        f"Recall:    {recall_value:.4f}"
    )

    print(
        f"F1:        {f1_value:.4f}"
    )


# ============================================================
# 18. MELHOR THRESHOLD PELO F1
# ============================================================

f1_scores = (
    2
    * precision_curve
    * recall_curve
    /
    (
        precision_curve
        + recall_curve
        + 1e-10
    )
)

melhor_indice = np.argmax(
    f1_scores
)

if melhor_indice < len(thresholds):

    melhor_threshold = thresholds[
        melhor_indice
    ]

else:

    melhor_threshold = 0.50


print("\n" + "=" * 60)
print("MELHOR THRESHOLD")
print("=" * 60)

print(
    "Threshold:",
    melhor_threshold
)

print(
    "F1:",
    f1_scores[melhor_indice]
)


# ============================================================
# 19. AVALIAÇÃO COM THRESHOLD
# ============================================================

y_pred_final = (
    prob_xgb >= melhor_threshold
).astype(int)

print("\n" + "=" * 60)
print("XGBOOST COM THRESHOLD AJUSTADO")
print("=" * 60)

print(
    classification_report(
        y_test,
        y_pred_final,
        digits=4
    )
)


# ============================================================
# 20. UNDERSAMPLING
# ============================================================

under = RandomUnderSampler(
    random_state=42
)

X_train_under, y_train_under = (
    under.fit_resample(
        X_train,
        y_train
    )
)

print("\n" + "=" * 60)
print("UNDERSAMPLING")
print("=" * 60)

print("\nAntes:")
print(
    y_train.value_counts()
)

print("\nDepois:")
print(
    y_train_under.value_counts()
)


rf_under = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_under.fit(
    X_train_under,
    y_train_under
)

pred_under = rf_under.predict(
    X_test
)

print("\nRandom Forest + Undersampling")

print(
    classification_report(
        y_test,
        pred_under,
        digits=4
    )
)


# ============================================================
# 21. OVERSAMPLING
# ============================================================

over = RandomOverSampler(
    random_state=42
)

X_train_over, y_train_over = (
    over.fit_resample(
        X_train,
        y_train
    )
)

print("\n" + "=" * 60)
print("OVERSAMPLING")
print("=" * 60)

print("\nAntes:")
print(
    y_train.value_counts()
)

print("\nDepois:")
print(
    y_train_over.value_counts()
)


rf_over = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_over.fit(
    X_train_over,
    y_train_over
)

pred_over = rf_over.predict(
    X_test
)

print("\nRandom Forest + Oversampling")

print(
    classification_report(
        y_test,
        pred_over,
        digits=4
    )
)


# ============================================================
# 22. FEATURE IMPORTANCE
# ============================================================

importancias = pd.Series(
    xgb_model.feature_importances_,
    index=X_train.columns
).sort_values(
    ascending=False
)

plt.figure(figsize=(10, 8))

importancias.head(15).sort_values().plot(
    kind="barh"
)

plt.title(
    "15 Principais Variáveis - XGBoost"
)

plt.xlabel(
    "Importância"
)

plt.show()


# ============================================================
# 23. SHAP - CORRIGIDO PARA VERSÕES ATUAIS
# ============================================================

# Usar somente uma amostra para acelerar

X_shap = X_test.sample(
    n=min(500, len(X_test)),
    random_state=42
)

# Criar explicador

explainer = shap.TreeExplainer(
    xgb_model
)

# Calcular valores SHAP

shap_values = explainer(
    X_shap
)


# ============================================================
# 24. SHAP SUMMARY PLOT
# ============================================================

plt.figure()

shap.plots.beeswarm(
    shap_values,
    max_display=15,
    show=False
)

plt.title(
    "SHAP - Principais Variáveis"
)

plt.tight_layout()

plt.show()


# ============================================================
# 25. IMPORTÂNCIA MÉDIA PELO SHAP
# ============================================================

shap_importance = pd.DataFrame({

    "feature":
        X_shap.columns,

    "importance":
        np.abs(
            shap_values.values
        ).mean(axis=0)
})

shap_importance = (
    shap_importance
    .sort_values(
        "importance",
        ascending=False
    )
)

print("\n" + "=" * 60)
print("TOP 15 VARIÁVEIS PELO SHAP")
print("=" * 60)

display(
    shap_importance.head(15)
)


# ============================================================
# 26. COMPARAÇÃO FINAL
# ============================================================

resultados = []

for nome, previsoes in modelos.items():

    resultados.append({

        "Modelo":
            nome,

        "Precision - Fraude":
            precision_score(
                y_test,
                previsoes,
                zero_division=0
            ),

        "Recall - Fraude":
            recall_score(
                y_test,
                previsoes,
                zero_division=0
            ),

        "F1 - Fraude":
            f1_score(
                y_test,
                previsoes,
                zero_division=0
            )
    })


resultados_df = pd.DataFrame(
    resultados
)

print("\n" + "=" * 60)
print("COMPARAÇÃO FINAL")
print("=" * 60)

display(
    resultados_df
)
```
